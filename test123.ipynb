```python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                    # 처리: 부가세 10%를 더한 금액 계산
print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
```

```python
# 상품 3개의 정가 입력 받기
onion_price = int(input("양파 1망 정가를 입력하세요 (원): "))
tomato_price = int(input("토마토 1팩 정가를 입력하세요 (원): "))
potato_price = int(input("감자 1박스 정가를 입력하세요 (원): "))

# 부가세(10%)가 포함된 최종 가격 계산
onion_total = int(onion_price * 1.1)
tomato_total = int(tomato_price * 1.1)
potato_total = int(potato_price * 1.1)

# 메뉴판 출력
print("="*40)
print(f" 양파 (1망)   : {onion_total:,} 원")
print(f" 토마토 (1팩) : {tomato_total:,} 원")
print(f" 감자 (1박스) : {potato_total:,} 원")
print("="*40)
```

```python
# 1. 기존 메뉴판 다시 보여주기
print("="*40)
print("        [ 부가세(10%) 포함 최종 메뉴판 ]        ")
print("="*40)
print(f" 양파 (1망)   : {onion_total:,} 원")
print(f" 토마토 (1팩) : {tomato_total:,} 원")
print(f" 21감자 (1박스) : {potato_total:,} 원")
print("="*40)
print()

# 2. 고객으로부터 상품 수량 입력 받기
potato_count = int(input("감자 몇 개 사시겠어요?: "))
tomato_count = int(input("토마토 몇 개 사시겠어요?: "))
onion_count = int(input("양파 몇 개 사시겠어요?: "))

# 3. 총 금액 계산
total_payment = (potato_total * potato_count) + (tomato_total * tomato_count) + (onion_total * onion_count)

# 4. 결과 출력
print("\n" + "-"*40)
print(f"주문 내역: 감자 {potato_count}개, 토마토 {tomato_count}개, 양파 {onion_count}개")
print(f"총 결제 금액은 {total_payment:,}원 입니다.")
print("-"*40)
```
